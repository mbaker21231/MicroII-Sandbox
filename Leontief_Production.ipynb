# Simple Leontief Example

Here is a simple example of a Leontief Production Problem.

Make sure the class knows exactly how to read the $A$ matrix: it is the amount of good $j$ needed to produce good $i$.


```python
import numpy as np

# Input coefficient matrix A (3x3)
A = np.array([
    [0.2, 0.1, 0.3],  # Inputs used from Agriculture
    [0.3, 0.4, 0.2],  # Inputs used from Manufacturing
    [0.1, 0.2, 0.2]   # Inputs used from Energy
])

# Final demand vector D (3x1)
D = np.array([80, 120, 60])

# Identity matrix I (3x3)
I = np.identity(3)

# Leontief inverse: (I - A)^(-1)
I_minus_A = I - A
inv = np.linalg.inv(I_minus_A)

# Compute gross output vector X
X = inv @ D

# Print results
goods = ["Agriculture", "Manufacturing", "Energy"]
print("Gross Output Needed:")
for i in range(3):
    print(f"{goods[i]}: {X[i]:.2f}")
```

## 🪑 Simple Production Optimization Problem

### 🧾 Problem Setup

A furniture factory produces two products: **tables** and **chairs**. Each product requires a certain amount of **labor** and **wood**, and each provides a specific **profit** per unit sold.

| Product      | Labor (hours per unit) | Wood (units per unit) | Profit per unit |
|--------------|------------------------|------------------------|------------------|
| **Table**    | 3                      | 4                      | \$70             |
| **Chair**    | 2                      | 3                      | \$50             |

The factory has the following resource constraints:
- A maximum of **240 hours** of labor available
- A maximum of **300 units** of wood available

---

### 🎯 Objective

Determine the optimal number of **tables** and **chairs** to produce in order to **maximize total profit**, subject to the available resources.

---

### 📐 Decision Variables

Let:
- \( x_1 \): number of **tables** to produce  
- \( x_2 \): number of **chairs** to produce

---

### 🔗 Mathematical Formulation

**Objective Function** (maximize profit):

$$
\text{Maximize } Z = 70x_1 + 50x_2
$$

**Subject to constraints:**

\[
\begin{aligned}
3x_1 + 2x_2 &\leq 240 \quad &\text{(Labor constraint)} \\
4x_1 + 3x_2 &\leq 300 \quad &\text{(Wood constraint)} \\
x_1, x_2 &\geq 0 \quad &\text{(Non-negativity)}
\end{aligned}
\]

```python
from scipy.optimize import linprog

# Coefficients for the objective function (negated for maximization)
c = [-70, -50]

# Coefficients for inequality constraints (Ax <= b)
A = [
    [3, 2],  # labor constraint
    [4, 3]   # wood constraint
]
b = [240, 300]

# Bounds for x1 and x2 (non-negative)
x_bounds = (0, None)
bounds = [x_bounds, x_bounds]

# Solve using Simplex method
result = linprog(c, A_ub=A, b_ub=b, bounds=bounds, method='highs')

# Output the result
if result.success:
    print("Optimal production plan:")
    print(f"Tables: {result.x[0]:.2f}")
    print(f"Chairs: {result.x[1]:.2f}")
    print(f"Maximum profit: ${-result.fun:.2f}")
else:
    print("Optimization failed:", result.message)
```
