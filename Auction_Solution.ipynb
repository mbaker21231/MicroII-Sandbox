# First-price sealed bid

- Players submit private bids
- Auctioneer awards the object to the highest bidder, who pays his or her bid for the object.

Suppose a particular player is considering bidding $b$. All other players follow a bid function $b^*(V)$. Hence, our player will win the auction if

$$
b>b^*(V)
$$

Which happens with probability

$$
P(b\geq b^*(V))\quad\rightarrow\quad P(\sigma(b)\geq v_{(n-1)}), \sigma(b)=b^{*-1}
$$

We then have:

$$
\rho(b)=F(\sigma(b))^{n-1}
$$

as the probability our player wins the auction. Note that $\sigma(b)$ gives, for any $b$, the value $v$ that leads to $b$ being bid. That is, $\sigma(b)$ is the inverse of the function $b^*$.

### Solution

The player's returns are:

$$
\rho(b)(v-b)
$$

which are maximized at:

$$
\rho'(b)(v-b)-\rho(b)=0
$$

From the above, we have:

$$
(n-1)F(\sigma(b))^{n-2}f(\sigma(b))\sigma'(b)(v-b)-F(\sigma(b))^{n-1}=0
$$

or

$$
(n-1)f(\sigma(b))\sigma'(b)(v-b)-F(\sigma(b))=0
$$

But now, since all the players are the same, they all must have the same strategy. Hence, $b=b(v)$, or $v=\sigma(b)$ in the above. Also, note that if $\sigma'(b)=\frac{1}{b'(v)}$, $v=\sigma(b)$. Then, we can write:

$$
(n-1)f(v)(v-b(v))-F(v)b'(v)=0
$$

In the uniform case, we have:

$$
(n-1)(v-b(v))-vb'(v)=0
$$

Bust out a diffyq solver...

```python
from sympy import symbols, Function, Derivative
from sympy.solvers.ode import dsolve
```

```python
n, v = symbols('n v', real=True, positive=True)
```

```python
b = Function('b')(v)
```

```python
deq = (n-1)*(v-b) - v*Derivative(b, v, 1)
```

```python
deq
```

```python
dsolve(deq, b)[1]
```

That works out, but we require that $C_1=0$; otherwise, the solution explodes as $v \rightarrow 0$
