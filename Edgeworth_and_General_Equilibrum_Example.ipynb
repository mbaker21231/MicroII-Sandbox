# General Equilibrium

Here is a basic two-consumer, two-good example where everything is plotted out in an Edgeworth Box.

# A Pure Exchange Economy

Illustration:

$$
u_1=x_{11}^\alpha x_{12}^{1-\alpha},\quad u_2=x_{21}^\beta x_{22}^{1-\beta}
$$

Maximize these subject to the budget constraint $w_1=p_1x_{11}+p_2x_{12}$, for example, using the Lagrangian:

$$
\mathbf{L} = x_{11}^\alpha x_{12}^{1-\alpha}+\lambda_1(w_1-p_1x_{11}-p_2x_{12})
$$
With first-order conditions:

$$
\frac{\partial L}{\partial x_{11}}=\alpha \left(\frac{x_{12}}{x_{11}}\right)^{1-\alpha}-\lambda_1p_1=0
$$

$$
\frac{\partial L}{\partial x_{12}}=(1-\alpha) \left(\frac{x_{11}}{x_{12}}\right)^{\alpha}-\lambda_1p_2=0
$$
Dividing the first equation by the second gives:

$$
\frac{\alpha}{1-\alpha}\frac{x_{12}}{x_{11}}=\frac{p_1}{p_2}
$$
So, $x_{12} = \frac{p_1}{p_2}\frac{1-\alpha}{\alpha}x_{11}$. Plug this into the budget constraint and solve, to get $x_{11}^*=\alpha w_1 / p_1$. In similar fashion we have demand functions:






This leads to demand functions:

$$
x_{11}^*=\frac{\alpha w_1}{p_1}, \quad x_{12}^*=\frac{(1-\alpha) w_1}{p_2}
$$

$$
x_{21}^*=\frac{\beta w_2}{p_1}, \quad x_{22}^*=\frac{(1-\beta) w_2}{p_2}
$$

Model is closed by modelling income as a function of endowments and prices (here, we part ways with Partial Equilibrium):

$$
w_1=\omega_{11}p_1+\omega_{12}p_2,\quad w_2=\omega_{21}p_1+\omega_{22}p_2
$$

Then:

$$
x_{11}^*=\frac{\alpha(\omega_{11}p_1+\omega_{12}p_2)}{p_1}, \quad x_{12}^*=\frac{(1-\alpha) (\omega_{11}p_1+\omega_{12}p_2)}{p_2}
$$

$$
x_{21}^*=\frac{\beta (\omega_{21}p_1+\omega_{22}p_2)}{p_1}, \quad x_{22}^*=\frac{(1-\beta) (\omega_{21}p_1+\omega_{22}p_2)}{p_2}
$$

We can then solve for equilibrium prices by requiring that demands equal endowments:

$$
x_{11}^*+x_{21}^*=\omega_{11}+\omega_{21},\quad x_{21}^*+x_{22}^*=\omega_{21}+\omega_{22}
$$

We will get into practical stuff with this in a minute. These last things, we added up, give us aggregate demands

# Income and preferences

Note in this case we have the budget lined determiend by something like:

$$
p_1x_{11} +p_2x_{12} = \omega_{11}p_1 + \omega_{12}p_2\quad\rightarrow\quad x_{12} = \omega_{11}p_1 + \omega_{12}p_2-\frac{x_{11}p_1}{p_2}
$$

So, we can normalize prices or think about a price ratio as the key thing. Common normalizations:

- fix one price at unity (numeraire, which we've already seen)
- simplex prices: $\sum_i p_i=1$

Exercise: if we use the first normalization, setting $p_2=1$, how does the price change the budget line? In this case, we have:

$$
x_{12} = (\omega_{11}-x_{11})p_1 + \omega_{12}
$$

Note how it goes through the endowment point. Always true, meaning that an agent can always choose to consume their endowment!

# Before that...three critical concepts

- **The core** - the set of economic outcomes that might be bargained to, independently of any sort of price system. They require the lack of "blocking" coalitions and therefore a sort of consensus.
- **The competitive (Walrasian) equilibrium.** Prices are taken as given, all markets clear in that demands exhaust endowments.
- **The (set of) Pareto-Optimal Allocations:** feasible allocations where no agent(s) can be made better off without making some worse off.

First, we note that any Pareto-Optimal allocation must be where the MRS's of the two agents are equal (why?). That is:

$$
\frac{\alpha y_1}{(1-\alpha)x_1}=\frac{\beta y_2}{(1-\beta)x_2}
$$

We will try to find something we can plot where $y_1=f(x_1)$. Pursuant to this, use $E_x, E_y$ as total endowment, where we can then substitute $x_2=E_x-x_1$, $y_2=E_y-y_1$.


In the Cobb-Douglas Example, where we have utilities as $x^\alpha y ^{1-\alpha}$, and $x^\beta y^{1-\beta}$
$$
y_1 = \frac{\beta(1-\alpha)x_1 E_y}{\alpha (1-\beta)(E_x-x_1)+\beta (1-\alpha)x_1}
$$

Indifference curves are given by, for example:

$$
e_{1x}^\alpha e_{1y}^{1-\alpha}=x^\alpha y ^{1-\alpha}
$$

So, we can plot these quite simply by observing that a monotone transform doesn't change anything so we get:

$$
y_1 = e_{1y}\left(\frac{e_{1x}}{x_1}\right)^\frac{\alpha}{1-\alpha}
$$

Now, since agent two is oriented at the upper right axis, we have to plot things a little differently. While we still have:

$$
y_2 = e_{2y}\left(\frac{e_{2x}}{x_2}\right)^\frac{\beta}{1-\beta}
$$

We need to define all these things in terms of $1$'s perspective. So, we have:

$$
E_y-y_1 = (E-e_{1y})\left(\frac{E_x-e_{1x}}{E_x-x_1}\right)^\frac{\alpha}{1-\alpha}
$$

Or

$$
y_1 = E_y-(E-e_{1y})\left(\frac{E_x-e_{1x}}{E_x-x_1}\right)^\frac{\alpha}{1-\alpha}
$$

The result is the edgeworth box:

### Walrasian Equilibrium

The aggregate demand function and individual demand functions. Note that the two people have the aggregate demand function. One can see from demand functions that only _relative_ prices matter. This is a consequence of homogeneity.

We can write aggregate demand for good one as:

$$
\alpha w_{11} + \beta w_{21} + (\alpha w_{12} + \beta w_{22})\frac{1-p}{p}
$$

Note that this equation is sufficient to describe an equilibrium $p$, where we want aggregate demand equal to aggregate supply, here $w_{11}+w_{21}$.

This equation could be solved for $p$ without reference to the demand for the other good - a version of Walras's Law. We find, in this particular case, that it is given by:

$$
\alpha w_{11} + \beta w_{21} + (\alpha w_{12} + \beta w_{22})\frac{1-p}{p}=w_{11}+w_{21}
$$

Solving this equation gives:

$$
p = \frac{\alpha w_{12}+\beta w_{22}}{\alpha (-w_{11}+ w_{12}) + \beta (-w_{21}+ w_{22})+w_{11}+w_{21} }
$$

```python
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import interact
```

```python
def EdgeworthBox(alpha=.5, beta=.5, e11=.5, e12=.5):

    Ex = 1
    Ey = 1

    xvals = np.arange(0.01, Ex, .01)

    pos = beta*(1-alpha)*xvals*Ey/( alpha*(1-beta)*(Ex-xvals)+beta*(1-alpha)*xvals)

    y1_ic = e12*(e11/xvals)**(alpha/(1-alpha))
    y1_ict = y1_ic[y1_ic<Ey]
    xvals_1t = xvals[y1_ic<Ey]

    y2_ic = Ey - (Ey - e12)*((Ex-e11)/(Ex-xvals))**(beta/(1-beta))
    y2_ict = y2_ic[y2_ic>0]
    xvals_2t = xvals[y2_ic>0]

    find_p1 = beta*(1-alpha)*xvals_1t*Ey/( alpha*(1-beta)*(Ex-xvals_1t)+beta*(1-alpha)*xvals_1t)
    find_p2 = beta*(1-alpha)*xvals_2t*Ey/( alpha*(1-beta)*(Ex-xvals_2t)+beta*(1-alpha)*xvals_2t)

    p1_cut = np.argmin(abs(find_p1 - y1_ict))
    p2_cut = np.argmin(abs(find_p2 - y2_ict))
    val_1 = find_p1[p1_cut]
    val_2 = find_p2[p2_cut]

    core_pos_picker = (pos >= val_1) & (pos <= val_2)

    cp1 = pos[core_pos_picker]
    xp1 = xvals[core_pos_picker]

    price = (alpha*e12 + beta*(1 - e12)) /(-alpha*e11 + alpha*e12 - beta*(1-e11) + beta*(1-e12) + 1)

    x11_sta = alpha*(e11*price + e12*(1-price))/price
    x12_sta = (1-alpha)*(e11*price + e12*(1-price))/(1-price)

    bline_x = np.arange(0, 1, .01)
    bline_y = e12 + price/(1-price)*e11 - price/(1-price)*bline_x



    plt.figure(figsize=(12,12))

    plt.title('Edgeworth Box Diagram')
    plt.ylabel('Endowment of Y')
    plt.xlabel('Endowment of X')
    plt.plot(xvals, pos, color='green')
    plt.plot(xp1, cp1, linewidth=4, color='black')
    plt.plot(xvals_1t, y1_ict, color='blue')
    plt.plot(xvals_2t, y2_ict, color='red')
    plt.legend(['Pareto Optima', 'Core Allocations', "u_1(x,y)=u", 'u_2(x,y)=u'])
    plt.scatter(e11, e12)
    plt.scatter(x11_sta, x12_sta, marker='X', s=200, color='red', linewidths=1)

    plt.plot(bline_x, bline_y)

    plt.text(1/2, -.3, ("Note: the total endowment has been normalized to unity.\n  \
        The alpha parameter is agent one's relative intensity of preference for good x,\n \
        while the beta parameter is agent two's relative intensity of preference for good x."), ha='center')
```

```python
f = interact(EdgeworthBox, alpha=(0.05,0.95,0.05),beta=(0.05,0.95,0.05),
             e11=(0,1,.01), e12=(0,1,.01))
```
