# Simple Public Goods Example

Let's take an example where individuals have utility:

$$
u_1 = a_1\ln(g_1+g_2) + ln(w_1-g_1)
$$
$$
u_2 = a_2\ln(g_1+g_2) + ln(w_2-g_2)
$$
These are utilities of two agents, where $G=g_1+g_2$ is the total amount of the public good. Note that it is nonrivalrous.

Let's solve first-order conditions and all that and see where that gets us. Note we will need to find a Nash equilibrium in contributions!

```python
from sympy import *
```

```python
a1, a2, g1, g2, w1, w2 = symbols('a1 a2 g1 g2 w1 w2')
```

```python
U1 = a1*log(g1+g2) + log(w1-g1)
U2 = a2*log(g1+g2) + log(w2-g2)
```

Finding the Nash equilibrium:

```python
f1 = diff(U1, g1)
f2 = diff(U2, g2)
```

```python
solns = solve([f1, f2], [g1, g2])
solns
```

A little difficult to see what's going on there. Let's try $a_1=a_2=1$. We have:

```python
solns[g1].subs({a1:1, a2:1}), solns[g2].subs({a1:1, a2:1})
```

```python
solns[g1].subs({a1:1, a2:1, w1:10, w2:10})
```

What about the social optimum? Well, let's try that out...

```python
f1 = diff(U1 + U2, g1)
f2 = diff(U1 + U2, g2)
solns = solve([f1, f2], [g1, g2])
solns
```

```python
solns[g1].subs({a1:1, a2:1}), solns[g2].subs({a1:1, a2:1})
```

```python
solns[g1].subs({a1:1, a2:1, w1:10, w2:10})
```

### Quasi-linear utility

What if we instead have quasi-linear utility? That is, what if:

$$
u_i = a_i \ln(g_1+g_2) - g_i, \quad i=1,2
$$

```python
u1 = a1*log(g1+g2) - g1
u2 = a2*log(g1+g2) - g2
f1 = diff(u1, g1)
f2 = diff(u2, g2)
solns = solve([f1, f2], [g1, g2])
solns
```

No nash equilibrium! Why is that the case? Well, look at the social optimum:

$$
U = (a_1+a_2)\ln (g_1 + g_2) - g_1-g_2
$$

We can say _how much_ of the good should be provided, but we can't say exactly by whom. What about a more explicit procedure...

# A larger example

What if we instead have a more explicit social procedure? We imagine that there are a continuum of types for which $a(i)=i$. Suppose moreover that we have our simple quasi-linear utility. Utility is:

$$
u_i = a_i \ln(G) - G
$$
So we can say that welfare is maximized where:

$$
\sum_i a_i\ln(G)di - G
$$
So $G^*=\sum_i a_i$. Now, suppose that agents each pay a share $\frac{1}{N}$ of the public good. We can then say that each agent's most preferred amount is:

$$
a_i \ln(G) - G/N
$$

So, if we differentiate the above, we get the following:
$$
\frac{a_i}{G}=\frac{1}{N}\quad\rightarrow\quad G^* = Na_i
$$

Why did we bother with this? Well the so-called _Median Voter Theorem_ says that, if agents vote pairwise over policies, the one preferred by the median voter will be the one invoked. Discussion question: how do

$$
\sum a_i
$$
and
$$
N a_{med}
$$
compare? We could get over or under provision, depending on the distribution of tastes!
