# Example from Baqaee et. al. (JPE, 2024)

The above is a paper that just appeared in the JPE by [Baqaee, Farhi, and Sangani](https://www.journals.uchicago.edu/doi/10.1086/727287). While the paper is all about Macro, it starts off with an interesting piece on Micro, which is their conception of a differentiated goods model with markup prices. I was intrigued by this, so I thought I would work it through.

Suppose our agent has income $wL + \Pi$, which is a fairly typical macroeconomic construction. In their paper, the agent has utility:

$$
u(Y,L) = \frac{Y^{1-\gamma}-1}{1-\gamma}-\frac{L^{1+\frac{1}{\zeta}}}{1+\frac{1}{\zeta}}
$$



But really what we care about is the way in which differentiated products are modeled. They say that the consumption bundle in the above is implicitly defined by 

$$
\int_0^1\Phi_0\left(\frac{y_\theta}{Y}\right)d\theta=1
$$

This was somewhat mysterious to me, so I had to delve into the literature a bit deeper. One useful reference is [Matsuyama and Ushchev](https://papers.ssrn.com/sol3/papers.cfm?abstract_id=3126492). Anyways, a good starting point is with the usual CES example, where we will find the indirect utility function and all of that that goes along with CES utility:

$$
U = \int_0^1 x_i^\sigma di
$$

Since the budget constraint is $h=\int_0^1p_ix_idi$, we have first order conditions:

$$
\sigma x_i^{\sigma-1}-\lambda p_i=0,\quad \sigma x_j^{\sigma-1}-\lambda p_j=0
$$
These imply that:

$$
\frac{x_i^{\sigma-1}}{p_i}=\frac{x_j^{\sigma-1}}{p_j}\quad\rightarrow\quad x_i = \left(\frac{p_j}{p_i}\right)^\frac{1}{1-\sigma}x_j
$$

When this is plugged back into the budget constraint, we get:

$$
\int_0^1 p_ix_idi = h \quad\rightarrow\quad \int_0^1 p_i \left(\frac{p_j}{p_i}\right)^\frac{1}{1-\sigma}x_j di
\quad\rightarrow\quad x_jp_j^\frac{1}{1-\sigma}\int_0^1 p_i^\frac{-\sigma}{1-\sigma}di=h
$$

This implies that the demand for variety $j$ is:

$$
x_j = \frac{h}{p_j^\frac{1}{1-\sigma}}\frac{1}{P}
$$
where $P=\int_0^1 p_i^\frac{-\sigma}{1-\sigma}di$

Incidentally, the _single aggregator demand functions_ introduced by Matsuyama and Ushchev are written as:

$$
x_i = \frac{h}{p_i}s_i\left(\frac{p_i}{A(p)}\right)
$$

The $s_i$ are share functions that sum to one. while $A(p)$ is a common price aggregator, which is a "linear-homogenous function of $p$ solving:

$$
\int s_i\left(\frac{p_i}{A}\right)=1
$$

which is exactly, evidently, what $\Phi$ above is. So, in our example, note that we can rewrite the demand for $x_j$ as:

$$
x_j = \frac{h}{p_j}\frac{1}{p_j^\frac{\sigma}{1-\sigma}}\frac{1}{P}
$$

Of course, this can be written as:

$$
x_j=\frac{h}{p_j}\frac{p_j^\frac{-\sigma}{1-\sigma}}{P}
$$

So, in the parlance of the paper, instead of getting $P$ by solving out the budget constraint, what we do is break up the demand $x_j$ into _shares_ and write:

$$
x_j = \frac{h}{p_j}s_j\left(\frac{p_j}{A}\right)
$$
In our case, once again, this becomes:

$$
x_j = \frac{h}{p_j}\left(\frac{p_j}{A}\right)^\frac{-\sigma}{1-\sigma}
$$

We then solve the "sum to one" condition for $A$ to get our price aggregator:

$$
\int_0^1\left(\frac{p_j}{A}\right)^\frac{-\sigma}{1-\sigma}dj=1
$$

In this case, we have:

$$
A = P = \left(\int_0^1 p^\frac{-\sigma}{1-\sigma}_j dj\right)^\frac{-(1-\sigma)}{\sigma}
$$

Okay. 

The consumer maximizes utility, then, by purchasing individual goods to create a $Y$ that is as large as possible given the budget constraint. So, we might equivalently think about how the consumer maximizes $Y$. 

### Special case

First, the usual assumption is that the above is governed by a power function. This is where we use $\Phi_0(\cdot)^\rho$, say. Then, we would have:

$$
\int_0^1\left(\frac{y_\theta}{Y}\right)^\rho d\theta=1\quad\rightarrow\quad Y^\rho = \int_0^1 y_\theta^\rho d\theta \quad\rightarrow\quad Y = \left(\int_0^1y_\theta^\rho d\theta\right)^\frac{1}{\rho}
$$





Which is a standard CES formulation. 

#### What do we get from the general formulation?

Let's solve the problem of maximizing $Y$ subject to a set of prices given a certain amount of income. With Lagrangians, we have:

$$
\max \frac{Y^{1-\gamma}-1}{1-\gamma}  + \lambda \left(1- \int_0^1\Phi_0\left(\frac{y_\theta}{Y}\right)d\theta\right) + \mu \left(I-\int_0^1p_\theta y_\theta d\theta\right)
$$


We first get:

$$
Y^{-\gamma}+ \lambda \int_0^1\frac{y_\theta}{Y^2}\Phi'_0\left(\frac{y_\theta}{Y}\right)=0
$$

With respect to any variety $i$, we have:

$$
-\lambda \frac{1}{Y}\Phi'_0\left(\frac{y_i}{Y}\right)-\mu p_i=0
$$

which is also true for variety $\theta$:

$$
-\lambda \frac{1}{Y}\Phi'_0\left(\frac{y_\theta}{Y}\right)-\mu p_\theta=0
$$
