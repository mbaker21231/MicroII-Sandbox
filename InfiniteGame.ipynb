# Basic Infinitely Repeated Labor Game

Idea: we have an employer who pays a high wage or a low wage $w_1$ or $w_0$. There is a worker who works hard $e=1$ or shirks $e=0$. Output is always low when the worker shirks, but is high with probability $p$ when the worker works hard.

The employer pays the worker a high wage after any period in which output was high, and a low wage in any period when output was low.

The employer wishes to get the employee to work hard always, by setting profit-maximizing (cost-minimizing) wages $w_1$ and $w_0$. The reservation wage is $\underline{u}$, the discount rage is $\beta$, and the costs of working hard are $c$.

Flow equations:

$$
W_{h1} = w_1-c + \beta pW_{h1} + \beta (1-p) W_{h0}
$$
$$
W_{h0} = w_0-c + \beta pW_{h1} + \beta (1-p) W_{h0}
$$
If the agent were to deviate once in the low state and then work hard:
$$
w_0+\beta W_{h0}
$$
And also in the high state:
$$
w_1 + \beta W_{h0}
$$
and the agent of course must not to want to deviate all the time:
$$
V_{0} = w_0 + \beta V_0
$$

Let's get all the closed-form value functions:

```python
from sympy import *
```

```python
w0, w1, Wh0, Wh1, p, beta, c = symbols('w_0 w_1 W_{h0} W_{h1} p beta c')
```

```python
flow1 = Wh1 - (w1-c+beta*p*Wh1+beta*(1-p)*Wh0)
flow0 = Wh0 - (w0-c+beta*p*Wh1+beta*(1-p)*Wh0)
```

```python
goodsolns = solve([flow1, flow0], [Wh1, Wh0])
goodsolns[Wh0].factor()
```

The last thing above is the return a worker gets when things went badly the previous period. We can add various constraints. First, the worker must prefer this to continuing to shirk, which are:



```python
V0 = w0/(1-beta)
V0
```

```python
diff = goodsolns[Wh0] - V0
diff
```

```python
w1hat = solve(diff, w1)[0]
w1hat
```

So we see that the wage in the good state must be equal to that above. Next, we have to ensure that the worker would prefer working hard in the low state to the reservation wage. Let's plug in the above and see what returns are, and they are:

```python
goodsolns[Wh0].subs(w1, w1hat).simplify()
```

Now, we see we have solved our problem. Set $w_0$ equal to reservation wage $\underline{u}$, and then set $w_1$ so that:

$$
w_1 = \underline{u} + \frac{c}{p\beta}
$$
