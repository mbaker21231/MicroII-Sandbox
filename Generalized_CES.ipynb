# Matsuyama and Ushschev

[Define some new types of demand functions](https://www.dropbox.com/scl/fi/x2sc12qui5e9jaxxcnk8o/JMP1_beyondCES.pdf?rlkey=t7o9y4n9xdpfab5ksa67n4bdg&e=1&dl=0), which make the usual CES setup more flexible. I'm going to experiment with this a bit here.

Their proposition one goes as follows:

Take a vector "share mapping" $s(\mathbf{z})=(s_1(z_1),\dots,s_n(z_n))^T$. Normalize it so that:
$$
\sum_i s_i(1) = 1
$$

Typically, the argument $z$ is a price divided by an aggregator; i.e., $z_i=\frac{p_i}{A}$

 Suppose it satisfies:

1. $s'_i(z_i)z_i < s_i(z_i)$
2. $s'_is'_j\geq 0$.

Note that the second property basically requires the functions to not switch signs, ever. This can be boiled down to some homothetic preference with $V(p,h)$ with an idea price index:

$$
V=\frac{h}{P(\mathbf{p})}
$$

with $P(\mathbf{p})$ given by:

$$
\ln P(\mathbf{p}) = \ln A(p) + \sum \int_{c_1}^{\frac{p_i}{A(p)}}\frac{s_i(\epsilon)}{\epsilon}d\epsilon
$$

Here, $A(P)$ is an aggregator that comes from solving the system:

$$
\sum_i^n s_i\left(\frac{p_i}{A}\right)=1
$$

### Test Drive 1

Just to test things out, let's suppose we take the share function to be of the constant form $\alpha_i$. Clearly, we need $\sum \alpha_i=1$, so that the constraint is obeyed. We then just take $A=1$, so we now have the price index as
$$
\ln P(\mathbf{p}) = \sum \int_{c_1}^{p_i} \frac{\alpha_i}{\epsilon}d\epsilon
$$
which becomes:
$$
\ln P(\mathbf{p}) = \sum \alpha_i \ln \epsilon|^{p_i}_{c_1}
$$
This then becomes:
$$
\ln P(\mathbf{p}) = \sum  \ln \epsilon^{\alpha_i}|^{p_i}_{c_1}
$$
Or
$$
\ln P(\mathbf{p}) =-\ln c_1+\sum \ln p_i^{\alpha_i}
$$

So, we find that:

$$
\mathbf{P} = \frac{\prod p_i^{\alpha_i}}{c_1}
$$
Which then implies that

$$
V(h,\mathbf{p}) = \frac{c_1h}{\frac{\prod p_i^{\alpha_i}}{c_1}}
$$
Which is exactly the indirect utility from the Cobb-Douglas production function.

### Questions and observations

1) Clearly, any function that is positive yet decreasing always satisfies the above conditions. What about a simple linear function?

2) THE CES function $s_i=\beta z_i^\sigma$ also does the trick, as noted in the paper. This really requires $\sigma<1$ (possibly negative too), because $s'_iz_i =\sigma \beta z_i^\sigma$. which is less than $\beta z_i^\sigma$ so long as $\sigma<1$.

### Upper bound

Suppose we write $s'z = s$. or $\frac{s'}{s}=z$. We clearly have $\ln s = \ln z + c$. Then, we get $s=e^c z$, which seemingly suggests that any function that grows at a rate that is less than linear will work.

Some candidates include logarithmic functions, any exponentiated function, iterations of logs and roots (i.e., $ln(ln(x))$ or $\sqrt{\ln(x)}$). And this doesn't include functions that can potentially go from positive to negative.
