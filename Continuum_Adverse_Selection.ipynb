# Simple continuum adverse selection model

Here, we take the skill level to be distributed uniformly on $a,b$, and we take the outside option to be given by the straight $w_0$ to $w_1$. We graph everything out accordingly.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define points a and b
a, b = 1, 4  # You can change these values

# Define x values between a and b
x = np.linspace(a, b, 100)

# Define y values for the three lines
y1 = x  # 45 degree line
y2_corrected = a + x / 2  # Conditional expectation line
y_new_corrected = np.linspace(0, 5, len(x))  # Reservation option line

# Find the intersection point of the red and green lines
intersection_idx = np.argmin(np.abs(y2_corrected - y_new_corrected))
intersection_x = x[intersection_idx]
intersection_y = y2_corrected[intersection_idx]  # or y_new_corrected[intersection_idx]

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x, y1, linestyle='dashed', label=r'45 degree line', color='blue')
plt.plot(x, y2_corrected, label=r'Conditional expectation', color='red')
plt.plot(x, y_new_corrected, label='Reservation option', color='green')

# Add vertical dotted line at the intersection
plt.axvline(x=intersection_x, linestyle='dotted', color='black', label='Intersection')

# Mark the endpoints and intersection point
plt.scatter([a, b], [a, b], color='blue', zorder=3)
plt.scatter([a, b], [a + a/2, b + b/2], color='red', zorder=3)
plt.scatter([a, b], [0, 5], color='green', zorder=3)
plt.scatter(intersection_x, intersection_y, color='black', zorder=4)  # Mark intersection

# Extend axes limits slightly beyond a and b
x_margin = (b - a) * 0.2  # 20% margin
plt.xlim(a - x_margin, b + x_margin)
plt.ylim(min(0, a + a/2) - x_margin, max(5, b + b/2) + x_margin)

# Labels and legend
plt.xlabel('x')
plt.ylabel('y')
plt.title('Simple Adverse Selection Model on a Continuum')
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```
