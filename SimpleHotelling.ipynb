```python
from sympy import *
```

So we can write equations:

$$
f(x)=x^2+\int_0^1 ldl
$$

```python
l, l1, l2 = symbols('l l_1 l_2')
```

```python
expr = integrate(l1-l,(l,0,l1))+integrate(l-l1,(l, l1,(l1+l2)/2))+integrate(l2-l,(l, (l1+l2)/2,l2))+integrate(l-l2,(l, l2, 1))
```

```python
expr
```

```python
simplify(expr)
```

```python
f1=diff(expr, l1)
f2=diff(expr, l2)
```

```python
solve([f1, f2], [l1, l2])
```
