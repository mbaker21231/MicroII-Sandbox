# Morality of Markets

My understanding of the paper by [Dewatripont and Tirole (2024)](https://www.journals.uchicago.edu/doi/epdf/10.1086/729445) which proffers evidence that markets are moral. Here is the basic idea behind the model:

- Suppliers determine a price $p_i$ and make an ethical or moral choice $a_i$. $a_i$ has an upper bound, $\overline{a}_i$, and it has a welfare impact $W_i(a_i)$, where $W_i'(a_i)$ is greater than zero. One interpretation offered is that $W_i(a_i)=-\psi(a_i)e$, where $\psi(a_i)$ is environmental damage and $e$ is an external cost. Evidently, $\psi$ is decreasing in $a_i$, and $a_i$ is then a level of abatement.
- Critical to note: $W_i(a_i)$ is in per-unit terms. 

So far, so good. here is where one has to be careful...

## Consumer attitudes

Consumers' cost or benefit of the action $a$ is given by $\phi(a_i)$, which means that they care about the net price

$$
\hat p_i = p_i + \phi(a_i)
$$

And we then have:

Consumers are:
- UPI (Unethical/present-biased/influenceable) when $\phi'>0$
- ethical when $\phi'<0$.

I have a hard time understanding $\phi'>0$. Indeed, DT say "modelling ethical consumers' behavior is straightfoward, $-\phi(a_i) \propto \alpha_c W_i(a_i)$ ... But they also say that UPI consumers are more interesting. These are people who _prefer lower moral standards_. The reasons offered for such behavior are not all that convincing to me. 

### Clearing up the markup

Let's think a little bit about the modified markup formula. As DT move along in their paper, they posit a standard demand function, where we have something like $q_i = D(\hat{\mathbf{p}})$. We could, alternatively, write $D(\hat{p}_i,\hat{p}_{-i})$. Some other assumptions:

- Products are substitutes (so demand is negative wrt own price and positive wrt other firms' prices).

#### Linear Demand, Monopolist

Let's think about a linear demand function and a monopolist and see what that would be like. Here, we would have something like:

$$
q_i = \alpha - \beta \hat{p}_i = \alpha - \beta (p_i+\theta(a_i))
$$
which really just becomes something like:
$$
q_i = \alpha - \beta \hat{p}_i = \alpha-\beta(\theta(a_i)) - \beta p_i
$$
So, separability seems to mean we have just an intercept shift. Note how UPI consumers then benefit from unethical behavior, while at the same time ethical behavior increases the demand for other consumers.

#### Exponential Demand, Monopolist

What if we used a demand function like $q_i=\alpha \hat{p}_i ^{-\beta}$? In this case, we have something like:

$$
q_i = q_i=\alpha (p_i+\phi(a_i)) ^{-\beta}
$$
Which is a little harder to parse. Let's come back to this when we bust out Roy's identity to take a deeper dive into these cases. 


#### Multinomial utility 

In the typical multinomial model, agents have utility from different alternatives, like:

$$
u_i = \kappa + \delta_i - p_i + \epsilon_i
$$
where $\alpha_i$ is some composite utility provided by the product. In the multinomial logit, we have the demand for good $i$ becoming something like:

$$
q_i = N\frac{e^{\frac{\delta_i-p_i}{\mu}}}{1+\sum_k e^{\frac{\delta_k-p_k}{\mu}}}
$$
Clearly, if we use $\hat{p}$ instead of just $p$, this is akin to modifying the $\delta_i$ in much the same way as we did before; it will now be 
$$
\hat{\delta}_i=\delta_i - \phi(a_i)
$$
Fair enough. 

Now, note in this multinomial setup that demand has the attribute that:

$$
\frac{\partial q_i}{\partial p_i}=-\frac{1}{\mu}N\frac{e^{\frac{\delta_i-p_i}{\mu}}}{1+\sum_k e^{\frac{\delta_k-p_k}{\mu}}}+\frac{1}{\mu}e^{\frac{\delta_i-p_i}{\mu}}N\frac{e^{\frac{\delta_i-p_i}{\mu}}}{\left(1+\sum_k e^{\frac{\delta_k-p_k}{\mu}}\right)^2}
$$

Which is commonly written as:

$$
\frac{\partial q_i}{\partial p_i}=-N\frac{1}{\mu}s_i(1-s_i)
$$
Which in turn gives a demand elasticity of:
$$
e_{ip} = \frac{p_i}{q_i}\frac{\partial q_i}{\partial p_i}=-\frac{1}{\mu}(1-s_i)p_i
$$

But note also that the elasticity of price with respect to ethical actions becomes:

$$
\frac{\partial q_i}{\partial a_i}=\frac{-\phi_i(a_i)}{\mu}N\frac{e^{\frac{\delta_i-p_i}{\mu}}}{1+\sum_k e^{\frac{\delta_k-p_k}{\mu}}}+\frac{\phi'(a_i)}{\mu}e^{\frac{\delta_i-p_i}{\mu}}N\frac{e^{\frac{\delta_i-p_i}{\mu}}}{\left(1+\sum_k e^{\frac{\delta_k-p_k}{\mu}}\right)^2}
$$

or

$$
e_{ia} = \frac{p_i}{q_i}\frac{\partial q_i}{\partial p_i}=-\frac{\phi'(a_i)}{\mu}(1-s_i)p_i
$$

## Costs

Costs are taken to be captured by:

$$
C_i=c_i(a_i)q_i + d_i(q_i)
$$

Costs can be either positive or negative in $a_i$. Just note that if $c'(a_i)>0$, firms then would behave unethically to keep costs low; that is, ethical behavior is costly. Of course, the interesting case is probably one in which ethical behavior is costly. One interesting case that is worth thinking about is one in which investors accept a reduction in return equal to $\alpha_IW_i(a_i)$, and for workers $\alpha_WW_i(a_i)$. Then, we would have:

$$
C_i=\gamma_i(a_i) - \alpha_IW_i(a_i)-\alpha_WW_i(a_i)
$$



So, in this case, moral actions may reduce the costs of doing business. 

## Supplier objectives
$$
V_i=[p_i-c_i(a_i)]D_i(\hat{p}) + \alpha_i \mathcal{W}_i(\hat p,a)
$$

Where $\mathcal{W}_i(p,a)$ denotes the firm's inherent ethics. What DT then do is go for a "broad bracketing" approach. This means that firms care about industry ethical activity:

$$
\mathcal{W}_i(\hat p,a) = \mathcal{W}(\hat p,a)=\mathcal{E}(\hat{p},a)=\sum W_j(a_j)D_j(\hat p)
$$

Don't forget that $\hat p$ is what matters for demand. 

## Irrelevance Result

If prices are flexible, supplier's behavior is independent of demand. Why?


$$
V_i=[p_i-c_i(a_i)]D_i(\hat{p}) + \alpha_i \mathcal{W}_i(\hat p,a)
$$

Note that this depends on actual price, not net prices, which figure into demand. If we wanted to substitute everything in as we might understand it, recall that $\hat p_i = p_i + \phi(a_i)$, and my inclination would be to substitute in for this everywhere, so we have: 

$$
V_i=[p_i-c_i(a_i)]D_i(p+\phi(a)) + \alpha_i \sum W_j(a_j)D_j(p+\phi(a))
$$

which can also be written as:

$$
V_i=[p_i-c_i(a_i)]D_i(p+\phi(a)) +\alpha_i \sum W_j(a_j)D_j(p+\phi(a))
$$





The first-order conditions are clearly:

$$
\frac{\partial V_i}{\partial p_i}=D'_i(p_i-c_i) +\sum_{j}\alpha_jW_j\frac{\partial D_j}{\partial p_i}+D_i
$$

$$
\frac{\partial V_i}{\partial a_i}=\left[D'_i(p_i-c_i)+\sum_{j}\alpha_jW_j\frac{\partial D_j}{\partial p_i}\right]\phi'(a_i)-D_ic'_i+\alpha_iW_i'D_i
$$



The irrelevance result stems from the following observation. Suppose we set the price condition equal to zero. We see that we can then solve this first first order condition collecting all the terms attached to $\phi'$ are equal to $D_i$. Hence:

$$
\frac{\partial V_i}{\partial a_i}=-D_i\phi'(a_i)-D_ic'_i+\alpha_iW_i'D_i=(\alpha_iW_i-\phi'(a_i)-c'_i)D_i=0
$$
