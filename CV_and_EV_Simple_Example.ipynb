## Simple Example of CV and EV

We will use the simple utility function $U(x,y)=x^\alpha y^{1-\alpha}$, for which we have $x^*=\frac{\alpha I}{p_x}$ and $y^*=\frac{(1-\alpha)I}{p_y}$. These are the Marshallian demands.

### Indirect utility

Substitute these back into the above to get:

$$
V=\alpha^\alpha (1-\alpha)^{1-\alpha}Ip_x^{-\alpha}p_y^{-(1-\alpha)}
$$

### Expenditure

That can be inverted to get

$$
E=\alpha^{-\alpha}(1-\alpha)^{-(1-\alpha)}Up_x^{\alpha}p_y^{(1-\alpha)}
$$

Note how expenditure is concave in prices and indirect utility is convex (decreasing) in prices. Just as a note, the hicksian demand is then:

$$
x^c = \alpha^{1-\alpha}(1-\alpha)^{-(1-\alpha)}Up_x^{\alpha-1}p_y^{(1-\alpha)}
$$

We can bring all of these into sympy as follows:

```python
from sympy import *

init_printing()

alpha, px, py, Y, U = symbols('alpha px py Y U')

V = alpha**alpha*(1-alpha)**(1-alpha) * Y * px**(-alpha)*py**(-(1-alpha))
E = alpha**(-alpha)*(1-alpha)**(-(1-alpha)) * U * px**(alpha)*py**(1-alpha)
V, E
```

Now, let's just get some fictional values to go with our example, so we have:

```python
vals = {alpha:1/4, px:1, py:1, Y:8}
V.subs(vals)
```

We can check that our expenditure function is correct as follows:

```python
vals2 = {alpha:1/4, px:1, py:1, U:V.subs(vals)}
E.subs(vals2)
```

What if the price of good $x$ rises to 2? What happens?

```python
vals2 = {alpha:1/4, px:2, py:1, U:V.subs(vals)}
vals2
```

```python
E.subs(vals2)
```

We see we need something like an additional 1.51 dollars to get back to where we were before. Note that this is less than how our expenditure might have been expected to rise if consumption bundles didn't change. That is, we were consuming:

$$
x = \frac{\alpha I }{p_x}= \frac{.25(8)}{1}=2
$$

So to sustain our old consumption, we would need an additional two dollars of income, but substitution allows partial insulation by substitution.

To get the Equivalent variation, we can start everything at the new values:

```python
vals2 = {alpha:1/4, px:2, py:1, Y:8}
V.subs(vals2)
```

```python
valse = {alpha:1/4, px:1, py:1, U:V.subs(vals2)}
E.subs(valse)
```
