# Multitask Principal Agent problem

We work through a simple two-task model. This follows the notes that are available online from S. Viswathan.

Suppose that output is:

$$
x_1=a_1 + \epsilon_1,\quad x_2 = a_2 + \epsilon_2
$$

The errors are normally distributed with means and covariances as follows:

$$
\boldsymbol{\epsilon} =
\begin{bmatrix}
0 \\
0
\end{bmatrix}, \quad
\boldsymbol{\Sigma} =
\begin{bmatrix}
\sigma_1^2 & \sigma_{12} \\
\sigma_{12} & \sigma_2^2
\end{bmatrix}
$$

Suppose that the agent gets a share $\alpha_1$, $\alpha_2$ of the output $x_1,x_2$. Then, the agent's utility can be written in certainty-equivalent form:

$$
\alpha_1 a_1 + \alpha a_2+\beta-C(a_1,a_2)-\frac{r}{2} \left(\alpha_1^2\sigma_1^2+2\alpha_1\alpha_2\sigma_{12}+\alpha_2^2\sigma_2^2\right)
$$

Note that rightmost term in the above is really just the matrix product $\alpha'\Sigma\alpha$. The principal's utility (the principal is risk neutral) is:

$$
a_1+a_2-\alpha_1a_1-\alpha_2a_2-\beta
$$

If we suppose that the agent has a reservation utility of 0, we then see that $\beta$ is determined as:

$$
\beta=-\alpha_1 a_1 - \alpha a_2+\beta+C(a_1,a_2)+\frac{r}{2} \left(\alpha_1^2\sigma_1^2+2\alpha_1\alpha_2\sigma_{12}+\alpha_2^2\sigma_2^2\right)
$$

Plug this value back into the principal's utility, and we have:

$$
a_1 + a_2 - C(a_1,a_2)-\frac{r}{2} \left(\alpha_1^2\sigma_1^2+2\alpha_1\alpha_2\sigma_{12}+\alpha_2^2\sigma_2^2\right)
$$

The principal doesn't care about risk, really, mind you. The principal only cares about it because it means a larger compensation must be given to the agent because of it!

Now, the above problem only has $a$'s and $\alpha$'s, so we are writing the problem as one where the principal wants higher effort, but needs to compensate the agent for the risk that coincident higher $\alpha$ creates. Note that from the agent's utility, we get:

$$
\alpha_1 - C'_1=0
$$
$$
\alpha_2-C'_2=0
$$
To maximize utility. Totally differentiate again to get:

$$
d \alpha _1 = C''_{11}da_1 + C''_{12}da_2
$$
$$
d \alpha_2 = C''_{12}da_1 + C''_{22}da_2
$$

Evidently:

$$
\frac{\partial \alpha_i}{\partial a_j}=C''_{ij}
$$

Now, we can differentiate the principal's utility:

$$
a_1 + a_2 - C(a_1,a_2)-\frac{r}{2} \left(\alpha_1^2\sigma_1^2+2\alpha_1\alpha_2\sigma_{12}+\alpha_2^2\sigma_2^2\right)
$$

To get:

$$
1  - C_1'-r\alpha_1\sigma_1^2C''_{11} - r\alpha_2\sigma_{12}C''_{11}-r\alpha_1\sigma_{12}C''_{12} - r\alpha_2\sigma_{2}^2C''_{12}=0
$$
$$
1 - C_2'-r\alpha_2\sigma_2^2C''_{22} - r\alpha_1\sigma_{12}C''_{22}-r\alpha_2\sigma_{12}C''_{12} - r\alpha_1\sigma_{1}^2C''_{12}=0
$$

But if we use $C_1'=\alpha_1$ and $C'_2=\alpha_2$ from the first-order conditions, we get these equations as:

$$
1  - \alpha_1-r\alpha_1\sigma_1^2C''_{11} - r\alpha_2\sigma_{12}C''_{11}-r\alpha_1\sigma_{12}C''_{12} - r\alpha_2\sigma_{2}^2C''_{12}=0
$$
$$
1 - \alpha_2-r\alpha_2\sigma_2^2C''_{22} - r\alpha_1\sigma_{12}C''_{22}-r\alpha_2\sigma_{12}C''_{12} - r\alpha_1\sigma_{1}^2C''_{12}=0
$$

So, we can solve these directly for the optimal shares $\alpha_1,\alpha_2$.

```python
from sympy import *
```

```python
alpha1,alpha2,r,sigma1,sigma2,sigma12, c11, c12, c22 = symbols("alpha_1 alpha_2 r sigma_1 sigma_2 sigma_{12} c_{11} c_{12} c_{22}")
```

```python
eq1 = 1 - alpha1 - r*alpha1*sigma1**2*c11 - r*alpha2*sigma12*c11 - r*alpha1*sigma12*c12 - r*alpha2*sigma2**2*c12
eq2 = 1 - alpha2 - r*alpha2*sigma2**2*c22 - r*alpha1*sigma12*c22 - r*alpha2*sigma12*c12 - r*alpha1*sigma1**2*c12
```

```python
solns = solve([eq1, eq2], [alpha1,alpha2])
```

```python
a1_sta = solns[alpha1].factor()
a2_sta = solns[alpha2].factor()
a1_sta
```

Not sure why that's not rendering correctly, but let's see what happens if we simplify this a bit. First, what happens if we let the tasks be uncorrelated and independent $c_{12}=0$. We get:

```python
a1_hat = a1_sta.subs({sigma12:0}).simplify()
a1_hat
```

Note that this doesn't make the shares independent. That's because both shares contribute to sharing risks. Now, What about the case in which we let $\sigma_2 \rightarrow \infty$? In this case, clearly, we require $\alpha_2$=0, as we can't shift even an iota of risk from $x_2$ onto the risk-averse agent. Now, we note that we can write the share for $a_1$ as:

```python
a1_flex = limit(a1_hat, sigma2, oo).simplify()
```

```python
a1_flex
```

If we now assume $c_{12}=0$, so that the tasks are cost-independent, we have:


```python
a1_flex.subs({c12:0}).simplify()
```

But back up for a minute. What if total effort is all that matters in the cost function, so $c(a_1,a_2) = c(a_1+a_2)$? Then, clearly, $c_{12}=c_{22}$, and note that we now have:

```python
a1_flex.subs({c12:c22})
```

The interpretation is that when one task is unobservable, it is not desirable to incentivise the other task. This might be a good explanation for why we don't very often observe high-powered incentives.
