# Solving a Second-Order Partial Differential Equation in SymPy

## The Heat Equation

The **two-dimensional heat equation** is a fundamental partial differential equation describing the distribution of heat (or diffusion of particles) over space and time:

\[
\frac{\partial u}{\partial t} = \alpha \left( \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2} \right)
\]

where:
- \( u(x, y, t) \) is the unknown function representing temperature distribution,
- \( \alpha \) is the thermal diffusivity (a constant),
- \( x, y \) represent spatial coordinates,
- \( t \) represents time.

This is a **second-order partial differential equation (PDE)** because it involves second-order derivatives with respect to spatial variables.

## Solving the PDE in SymPy

We can use `sympy` to solve this PDE symbolically. The following code defines and attempts to solve the equation:

```python
from sympy import *

# Define the variables
x, y, t, alpha = symbols('x y t alpha')
u = Function('u')(x, y, t)

# Define the PDE (heat equation)
pde = Eq(sp.diff(u, t), alpha * (diff(u, x, x) + diff(u, y, y)))

# Solve the PDE
solution = dsolve(pde)

# Display the solution
solution
```
