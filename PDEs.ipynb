# Solving a Second-Order Partial Differential Equation in SymPy

## The Heat Equation

The **one-dimensional heat equation** is a fundamental partial differential equation describing the distribution of heat (or diffusion of particles) over space and time:

$$
\frac{\partial u}{\partial t} = \alpha \frac{\partial^2 u}{\partial x^2}
$$

where:
- $ u(x, t) $ is the unknown function representing temperature distribution,
- $ \alpha $ is the thermal diffusivity (a constant),
- $ x, y $ represent spatial coordinates,
- $ t $ represents time.

This is a **second-order partial differential equation (PDE)** because it involves second-order derivatives with respect to spatial variables.

## Solving the PDE in SymPy

We can use `sympy` to solve this PDE symbolically. The following code defines and attempts to solve the equation:

```python
import sympy as sp

# Define variables
x, t, alpha = sp.symbols('x t alpha')
U = sp.Function('U')(x, t)

# Define the PDE
pde = sp.Eq(sp.diff(U, t), alpha * sp.diff(U, x, x))

# Assume a separable solution: U(x,t) = X(x)T(t)
X = sp.Function('X')(x)
T = sp.Function('T')(t)
U_sep = X * T  # Product of separate functions

# Substitute into the PDE
pde_sep = pde.subs(U, U_sep).doit()

# Extract left-hand and right-hand sides
lhs = pde_sep.lhs
rhs = pde_sep.rhs

# Separate variables by dividing both sides by (X * T)
sep_eq = sp.Eq(lhs / (X * T), rhs / (X * T))

# Simplify the equation
sep_eq = sp.simplify(sep_eq)

# Introduce the separation constant (-lambda)
lambda_ = sp.symbols('lambda')
eq1 = sp.Eq(sep_eq.lhs, -lambda_)

# Solve for spatial part X(x)
eq_x = sp.Eq(sp.diff(X, x, x), (lambda_ / alpha) * X)
X_sol = sp.dsolve(eq_x, X)

# Solve for time part T(t)
eq_t = sp.Eq(sp.diff(T, t), -lambda_ * T)
T_sol = sp.dsolve(eq_t, T)

# Display solutions
X_sol, T_sol
```

Better yet, here is a way to print the solutions so they look a little nicer...

```python
from IPython.display import display, Math  # Import for LaTeX rendering

display(Math(r"\text{Equation after separation:}"))
display(Math(sp.latex(eq1)))

display(Math(r"\text{Solution for } X(x):"))
display(Math(sp.latex(X_sol)))

display(Math(r"\text{Solution for } T(t):"))
display(Math(sp.latex(T_sol)))
```

Here is another example, where we are solving a standard differential equation for the present value of something that evolves according to a geometric brownian motion and gives a return proportional to $x$:

```python
# Define variables
x, sigma, r, delta, pi = sp.symbols('x sigma r delta pi', real=True, positive=True)
F = sp.Function('F')(x)  # Solution function

# Define the ODE
ode = sp.Eq(
    (1/2) * sigma**2 * x**2 * sp.diff(F, x, x) + (r - delta) * x * sp.diff(F, x) - r * F + pi*x,
    0
)

# Solve the ODE
solution = sp.dsolve(ode, F)

# Display the solution
sp.pprint(solution, use_unicode=True)
```
