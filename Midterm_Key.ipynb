# Answer key - Midterm Exam

Consider a game in which players have payoffs:
$$
U_i=-(x_i-t_i)^2 - (x_i-x_{-i})^2
$$
Here $x_i$ is the action chosen by the player, $t_i$ is a given target a player wants to be close to, and $x_{-i}$ is some measure of what other players do.

Suppose that there are only two players, $i=0,1$. Suppose also that $t_i=i$, so player 0's target is 0, and player 1's target is 1. Find the Nash equilibrium of the game, and graph the reaction functions of each player, showing the equilibrium.

### Answer

Differentiate first order conditions with respect to own actions, set result to zero, and solve the resulting system simultaneously.


```python
from sympy import *
```

```python
t1, t2, x1, x2 = symbols('t_1 t_2 x_1 x_2')
```

```python
e1 = -(x1-t1)**2 - (x1-x2)**2
e2 = -(x2-t2)**2 - (x2-x1)**2
```

```python
f1 = diff(e1.subs({t1:0}), x1)
f2 = diff(e2.subs({t2:1}), x2)
print(f1, f2)
```

```python
solve([f1, f2], [x1, x2])
```

Suppose instead that the game is played sequentially with player 0 moving first. Find the new Nash equilibrium? First, solve the reaction function of player 2:

Next problem is to suppose that player 0 moves first. In this event, we need to use the reaction function for player 1 in player 0's problem.  

```python
e1mf = e1.subs({x2:solve(f2,x2)[0], t1:0})
```

```python
e1mf
```

```python
f1mf = diff(e1mf, x1)
f1mf
```

```python
solve([f1mf, f2], [x1, x2])
```

Socially optimal actions require maximizing the sum of the two actions:

```python
f1so = diff(e1 + e2, x1)
f2so = diff(e1 + e2, x2)
```

```python
solve([f1so.subs({t1:0}), f2so.subs({t2:1})], [x1, x2])
```
