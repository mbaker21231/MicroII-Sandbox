# Answer key - Midterm Exam

Consider a game in which players have payoffs:
$$
U_i=-(x_i-t_i)^2 - (x_i-x_{-i})^2
$$
Here $x_i$ is the action chosen by the player, $t_i$ is a given target a player wants to be close to, and $x_{-i}$ is some measure of what other players do.

Suppose that there are only two players, $i=0,1$. Suppose also that $t_i=i$, so player 0's target is 0, and player 1's target is 1. Find the Nash equilibrium of the game, and graph the reaction functions of each player, showing the equilibrium.

### Answer

Differentiate first order conditions with respect to own actions, set result to zero, and solve the resulting system simultaneously.


```python
from sympy import *
```

```python
t1, t2, x1, x2 = symbols('t_1 t_2 x_1 x_2')
```

```python
e1 = -(x1-t1)**2 - (x1-x2)**2
e2 = -(x2-t2)**2 - (x2-x1)**2
```

```python
f1 = diff(e1.subs({t1:0}), x1)
f2 = diff(e2.subs({t2:1}), x2)
print(f1, f2)
```

```python
solve([f1, f2], [x1, x2])
```

Suppose instead that the game is played sequentially with player 0 moving first. Find the new Nash equilibrium? First, solve the reaction function of player 2:

Next problem is to suppose that player 0 moves first. In this event, we need to use the reaction function for player 1 in player 0's problem.  

```python
e1mf = e1.subs({x2:solve(f2,x2)[0], t1:0})
```

```python
e1mf
```

```python
f1mf = diff(e1mf, x1)
f1mf
```

```python
solve([f1mf, f2], [x1, x2])
```

Socially optimal actions require maximizing the sum of the two actions:

```python
f1so = diff(e1 + e2, x1)
f2so = diff(e1 + e2, x2)
```

```python
solve([f1so.subs({t1:0}), f2so.subs({t2:1})], [x1, x2])
```

A principal wishes to motivate an agent. The principal pays the agent a base wage of $w$, and if the outcome is good, the principal pays the agent a bonus $b$. There are two possible outcomes: a good outcome, which creates profits of $\pi$, and a bad outcome, which creates profits of $0$. The agent exerts effort $a$. The probability of a good outcome is $p_g=k\sqrt{a}$, where $k$ is a constant. The costs of effort are simply $a$.   

The agent dislikes varying wages, which we approximate with a utility function where $w_g$ denotes total compensation in the good state and $w_b$ total compensation in the bad state:

$$
Eu = p_gw_g + p_bw_b - r\frac{1}{2}(w_g-w_b)^2 - a
$$

### Question:

Write expected utility in terms of the bonus $b$ and the base wage, $w$. Find the agent's optimal choice of effort as a function of the payment parameters $w$ and $b$.

Fairly simple substitution, so we have:
$$
Eu = k\sqrt{a}b + w - r\frac{1}{2}b^2 - a
$$

```python
k, a, b, w, r, pi, u = symbols('k a b w r pi u')
```

```python
Eu = k*sqrt(a)*b +w - r*b**2/2 - a
```

```python
fa = diff(Eu, a)
```

```python
asta = solve(fa, a)
asta[0]
```

Principal wishes to maximize:

$$
p_g(\pi-b) - w
$$

```python
simplify(k*sqrt(asta[0])*(pi-b) - w)
```

```python
PP = k**2*b*(pi-b)/2 - w
PP
```

Agent's utility is:

```python
Eurf = Eu.subs({a:asta[0]}).simplify()
Eurf
```

```python
Eurf = b**2*k**2/4 - b**2*r/2 + w
```

```python
Eurf
```

We have to set this equal to reservation utility to get equilibrium w. So, for the reduced-form base wage, we have:

```python
wrf = solve(Eurf-u, w)[0]
wrf
```

Principal's utility is:

```python
PPrf = PP.subs({w:wrf})
PPrf
```

```python
bsta=solve(diff(PPrf, b), b)[0]
bsta
```

```python
wrf.subs({b:bsta}).factor()
```
